```python
def generate_key_square(key):
    key = key.replace(" ", "").upper()
    key_square = []
    # Use a set to keep track of characters already added to the key
    # square
    seen_chars = set()
    # Add key characters to the key square
    for char in key:
        if char == 'J':
            char = 'I' # J is usually treated as I in Playfair
        if char not in seen_chars:
            key_square.append(char)
            seen_chars.add(char)
    # Add remaining alphabet characters (A-Z, excluding J, in order)
    for char_code in range(ord('A'), ord('Z') + 1):
        char = chr(char_code)
        if char == 'J':
            continue
        if char not in seen_chars:
            key_square.append(char)
            seen_chars.add(char)
    # Reshape into a 5x5 matrix
    matrix = []
    for i in range(0, 25, 5):
        matrix.append(key_square[i:i+5])
    return matrix
def prepare_plaintext(plaintext):
    plaintext = plaintext.replace(" ", "").upper()
    processed_text = []
    i = 0
    while i < len(plaintext):
        char1 = plaintext[i]
        if char1 == 'J':
            char1 = 'I'
        if i + 1 == len(plaintext): # Last character
            processed_text.append(char1)
            processed_text.append('X') # Pad with 'X' if odd length
            break

        char2 = plaintext[i+1]
        if char2 == 'J':
            char2 = 'I'
        if char1 == char2:
            processed_text.append(char1)
            processed_text.append('X')
            i += 1
        else:
            processed_text.append(char1)
            processed_text.append(char2)
            i += 2
    return "".join(processed_text)
def find_char_position(matrix, char):
    for r, row in enumerate(matrix):
        for c, val in enumerate(row):
            if val == char:
                return r, c
    return -1, -1 # Should not happen if char is in key square
def encrypt_digraph(matrix, digraph):
    char1, char2 = digraph[0], digraph[1]
    r1, c1 = find_char_position(matrix, char1)
    r2, c2 = find_char_position(matrix, char2)
    if r1 == r2: # Same row
        return matrix[r1][(c1 + 1) % 5] + matrix[r2][(c2 + 1) % 5]
    elif c1 == c2: # Same column
        return matrix[(r1 + 1) % 5][c1] + matrix[(r2 + 1) % 5][c2]
    else: # Rectangle
        return matrix[r1][c2] + matrix[r2][c1]
def playfair_encrypt(plaintext, key):
    key_square = generate_key_square(key)
    prepared_text = prepare_plaintext(plaintext)
    # Ensure prepared_text has even length after processing
    if len(prepared_text) % 2 != 0:
        prepared_text += 'X'
    ciphertext = []
    for i in range(0, len(prepared_text), 2):

        digraph = prepared_text[i:i+2]
        ciphertext.append(encrypt_digraph(key_square, digraph))
    return "".join(ciphertext)
# Example Usage:
key = input("Enter the key: ")
plaintext = input("Enter the plaintext: ")
ciphertext = playfair_encrypt(plaintext, key)
print(f"Key Square:")
matrix = generate_key_square(key)
for row in matrix:
    print(row)
print(f"Prepared Plaintext: {prepare_plaintext(plaintext)}")
print(f"Encrypted Text: {ciphertext}")
```
